## Bài 9. Gradient Descent tìm hệ số đa thức

```python
import numpy as np
import torch
polynomial = np.poly1d([1, 2, 3])
N = 20
X = np.random.randn(N, 1) * 5
Y = polynomial(X)
XX = np.hstack([
X * X,
X,
np.ones_like(X)
])
w = torch.randn(3, 1, requires_grad=True)
x = torch.tensor(XX, dtype=torch.float32)
y = torch.tensor(Y, dtype=torch.float32)
optimizer = torch.optim.NAdam([w], lr=0.01)
print("Initial coefficients:")
print(w)
for epoch in range(1000):
    optimizer.zero_grad()
    y_pred = x @ w
    mse = torch.mean(torch.square(y - y_pred))
    mse.backward()
    optimizer.step()
print("Learned coefficients:")
print(w)
```

## Bài 10. Giải hệ phương trình bằng Autograd

```python
A = torch.tensor(random.random(), requires_grad=True)
B = torch.tensor(random.random(), requires_grad=True)
C = torch.tensor(random.random(), requires_grad=True)
D = torch.tensor(random.random(), requires_grad=True)
EPOCHS = 2000
optimizer = torch.optim.NAdam([A, B, C, D], lr=0.01)
for _ in range(EPOCHS):
    y1 = A + B - 9
    y2 = C - D - 1
    y3 = A + C - 8
    y4 = B - D - 2
    sqerr = y1*y1 + y2*y2 + y3*y3 + y4*y4
    optimizer.zero_grad()
    sqerr.backward()
    optimizer.step()
print("A =", A)
print("B =", B)
print("C =", C)
print("D =", D)
```

## Bài 11. Viết chương trình PyTorch để thực hiện các yêu cầu sau:
* Tạo Tensor X có 24 phần tử và reshape lần lượt thành 4 × 6, 2 × 12 và 2 × 3 × 4.
* Tạo hai Tensor A, B kích thước 3 × 4; tính A+B, A*B, A/B, trung bình theo cột và độ lệch chuẩn theo cột.
* Cho y = 2x2 + 5x + 3. Dùng autograd tính dy/dx tại x = 2 và kiểm tra bằng phép tính thủ công.
* Thay đa thức trong Bài 9 bằng y = 2x2 - 3x + 5; huấn luyện và cho biết các hệ số tìm được.

```python
import torch

print("--- YÊU CẦU 1 ---")
X = torch.arange(24)
X_4x6 = X.reshape(4, 6)
X_2x12 = X.reshape(2, 12)
X_2x3x4 = X.reshape(2, 3, 4)

print("Kích thước 4x6:\n", X_4x6)
print("Kích thước 2x12:\n", X_2x12)
print("Kích thước 2x3x4:\n", X_2x3x4)

print("\n--- YÊU CẦU 2 ---")
torch.manual_seed(42)
A = torch.randn(3, 4)
B = torch.randn(3, 4)

print("Tensor A:\n", A)
print("Tensor B:\n", B)
print("A + B:\n", A + B)
print("A * B (Element-wise):\n", A * B)
print("A / B (Element-wise):\n", A / B)

print("Trung bình theo cột của A:", torch.mean(A, dim=0))
print("Độ lệch chuẩn theo cột của A:", torch.std(A, dim=0))

print("\n--- YÊU CẦU 3 ---")
x = torch.tensor(2.0, requires_grad=True)
y = 2 * x**2 + 5 * x + 3
y.backward()

print("Tính bằng PyTorch autograd (dy/dx tại x=2):", x.grad.item())

print("\n--- YÊU CẦU 4 ---")
np_x = torch.randn(100, 1) * 5
y_true = 2 * (np_x ** 2) - 3 * np_x + 5

XX = torch.cat([np_x**2, np_x, torch.ones_like(np_x)], dim=1)

w = torch.randn(3, 1, requires_grad=True)
optimizer = torch.optim.NAdam([w], lr=0.01)

for epoch in range(2000):
    optimizer.zero_grad()
    y_pred = XX @ w
    mse = torch.mean((y_true - y_pred) ** 2)
    mse.backward()
    optimizer.step()

learned_w = w.detach().numpy().ravel()
print(f"Hệ số tìm được (a, b, c): a = {learned_w[0]:.2f}, b = {learned_w[1]:.2f}, c = {learned_w[2]:.2f}")
print("Hệ số lý thuyết ban đầu : a = 2.00, b = -3.00, c = 5.00")
```
